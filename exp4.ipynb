```python
import numpy as np
import pandas as pd

# Set seed for reproducibility
np.random.seed(17)
n = 20

# --- For Duplicating numbers ---
ID = np.arange(1, n + 1)
# Creating a duplicate by assigning the value of index -3 to index -2
ID[-2] = ID[-3]

print(ID)
# Note: The handwritten print(ID[-27]) would throw an error if n=20 
# as it's out of bounds, but following the notes:
# print(ID[-2]) 
# print(ID[-8])

# --- For generating Random names from given list ---
names_list = ["Komal", "simar", "jatin", "mahi", "vivaan", "Riya", "Vihaan"]
Names = np.random.choice(names_list, n)
print(Names)

# --- For generating Random age using randint function ---
# Note: The notes show two ways, here is the second implementation with NaNs
Age_list = np.random.randint(21, 55, size=n).astype(float)
# Introducing missing values (NaN) at specific indices
Age_list[[1, 6]] = np.nan
Age = Age_list
print(Age)

# --- For generating Random genders using p variable (probability) ---
Genders = np.random.choice(["M", "F", np.nan], size=n, p=[0.50, 0.45, 0.05])
print(Genders)

# --- Generating Salary with randint and astype function ---
Salary_list = np.random.randint(21000, 55000, size=n).astype(float)
# Introducing missing values (NaN) at specific indices
Salary_list[[2, 8]] = np.nan
Salary = Salary_list
print(Salary)

# --- Generating dataframe ---
data = pd.DataFrame({
    "ID": ID,
    "Names": Names,
    "Gender": Genders,
    "Age": Age,
    "Salary": Salary
})

print(data)
```
