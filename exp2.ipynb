```python
# Creating a dictionary
student = {
    "Name": "Simar",
    "Rollno": 3541,
    "Subject": "Machine Learning",
    "Marks": 95
}

# Display Dictionary
print(student)

# Access Values Using keys
print("Name:", student["Name"])
print("Rollno:", student["Rollno"])
print("Subject:", student["Subject"])

# Update Values
student["Marks"] = 90
print("Updated marks:", student["Marks"])
print(student)

# Add new Key-Value pair
student["Grade"] = "A"
print(student)

# Display all keys
print("All the keys are:")
print("Keys:", student.keys())

# Display all Values
print("All the Values of dictionary are:")
print("Values:", student.values())

# Display Key-Value pair
print("All key-Value pair in dictionary are:")
print("Items:", student.items())

# --- Put data into the dictionary (Nested Dictionary Example) ---

names = ["Simar", "Amit", "Suman", "Ravi", "Sonia"]
rollno = [1, 2, 3, 4, 5]
subjects = ["Maths", "Java", "DBMS", "ML", "CN"]
marks = [85, 90, 78, 92, 88]
data = {}

for i in range(len(names)):
    data[rollno[i]] = {
        "name": names[i],
        "Subject": subjects[i],
        "marks": marks[i]
    }

# Display nested dictionary data
for roll, info in data.items():
    print("Roll No:", roll)
    print("Name:", info["name"])
    print("Subject:", info["Subject"])
    print("Marks:", info["marks"])
    print("-" * 25)

# --- To create a dictionary using two different lists ---

data_dict = dict()
print(data_dict)

names_list = ["Simar", "Amit", "Suman", "Ravi", "Sonia"]
rollno_list = [1, 2, 3, 4, 5]

keys = ['names', 'rollno']
values = [names_list, rollno_list]

for i in range(len(keys)):
    data_dict[keys[i]] = values[i]

print(data_dict)
```
